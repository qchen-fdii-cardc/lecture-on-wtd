```python
import cadquery as cq

# Define parameters for the model
length = 100
width = 50
height = 30
hole_diameter = 40

model = (
    cq.Workplane("XY")
    .box(length, width, height)
    .faces(">Z").workplane()
    .circle(hole_diameter / 2)
    .cutThruAll()
)

# Let's try displaying it via an Assembly object, which can be more robust.
# We will also give it an explicit color.
assembly = cq.Assembly()
assembly.add(model, color=cq.Color("lightgray"))

# Display the assembly
assembly
```
