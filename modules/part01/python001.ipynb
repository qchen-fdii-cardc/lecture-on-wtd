### NumPy: 数值 Python

```python
# 按照惯例，NumPy 被导入并简写为 'np'
import numpy as np

# 从 Python 列表创建 NumPy 数组
my_list = [1, 2, 3, 4, 5]
np_array = np.array(my_list)

print(f"这是一个 NumPy 数组: {np_array}")
print(f"它的类型是: {type(np_array)}")

# NumPy 数组允许逐元素操作
doubled_array = np_array * 2
print(f"数组元素翻倍: {doubled_array}")

# 创建一个数字范围的数组
range_array = np.arange(0, 10, 2) # 开始, 结束 (不包含), 步长
print(f"从范围创建的数组: {range_array}")

# 创建一个多维数组 (矩阵)
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("\\n一个 2x3 矩阵:")
print(matrix)
print(f"矩阵的形状: {matrix.shape}")
```

### Pandas: 数据分析库

Pandas 提供了高性能、易于使用的数据结构和数据分析工具。Pandas 中的主要数据结构是 `Series` (一维) 和 `DataFrame` (二维)。DataFrame 就像一个电子表格或 SQL 表，是 Pandas 中最常用的对象。

```python
# 按照惯例，pandas 被导入并简写为 'pd'
import pandas as pd

# 从字典创建 DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df = pd.DataFrame(data)

# 显示 DataFrame
print("一个 pandas DataFrame:")
print(df)

# 获取数据摘要
print("\\nDataFrame 信息:")
df.info()

# 选择单列 (这是一个 pandas Series)
print("\\n选择 'Name' 列:")
print(df['Name'])

# 根据条件筛选数据
print("\\n年龄大于 30 的人:")
print(df[df['Age'] > 30])
```

### Matplotlib: 绘图和可视化

Matplotlib 是一个用于在 Python 中创建静态、动画和交互式可视化的综合库。它具有高度可定制性，并提供了种类繁多的绘图和图表。

```python
# 按照惯例，pyplot 模块被导入并简写为 'plt'
import matplotlib.pyplot as plt
import numpy as np


# 简单的线图
x_values = np.linspace(0, 10, 100) # 从 0 到 10 的 100 个点
y_values = np.sin(x_values)

plt.figure(figsize=(8, 4)) # 创建一个特定大小的图形
plt.plot(x_values, y_values)
plt.title('Sine Wave')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show() # 显示绘图

# 散点图
x_random = np.random.rand(50) * 10
y_random = np.random.rand(50) * 10

plt.figure(figsize=(8, 4))
plt.scatter(x_random, y_random, c='red', marker='x')
plt.title('Random Scatter Plot')
plt.xlabel('Random X')
plt.ylabel('Random Y')
plt.grid(True)
plt.show()
```

### SciPy: 科学计算

SciPy 建立在 NumPy 之上，为数值积分、插值、优化、线性代数和统计等任务提供了大量用户友好且高效的数值例程。

让我们看一个简单的例子：使用 `scipy.optimize` 寻找函数的最小值。

```python
from scipy.optimize import minimize
import numpy as np
import matplotlib.pyplot as plt

# 定义一个我们想要最小化的函数
# 让我们使用一个简单的二次函数: f(x) = (x - 3)^2
def objective_function(x):
    return (x - 3)**2

# 我们需要一个最小值的初始猜测值
initial_guess = 0.0

# 调用 minimize 函数
result = minimize(objective_function, initial_guess)

# result 对象包含有关优化的信息
if result.success:
    found_minimum = result.x[0]
    print(f"函数的最小值位于 x = {found_minimum:.4f}")
else:
    print("优化失败。")

# 我们可以绘制这个来验证
x_vals = np.linspace(0, 6, 100)
y_vals = objective_function(x_vals)

plt.figure(figsize=(8, 4))
plt.plot(x_vals, y_vals, label='f(x) = (x - 3)^2')
plt.plot(found_minimum, objective_function(found_minimum), 'ro', label=f'Minimum at x={found_minimum:.2f}')
plt.title('using SciPy Minimize a function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

### CoolProp: 热物理性质

CoolProp 是一个重要的开源库，为工程师和科学家提供各种流体和混合物的精确热物理性质数据。它广泛应用于热力学、传热和流体动力学模拟。

使用 CoolProp，您可以轻松计算水、制冷剂和工业气体等物质的密度、粘度、焓、熵等性质。


```python
# 现在让我们使用 CoolProp 获取水的一些性质。
import CoolProp.CoolProp as CP

# 定义流体的状态
# 我们将使用大气压 (101325 帕斯卡) 和 100°C (373.15 开尔文) 的水
pressure = 101325  # Pa
temperature = 373.15 # K
fluid = 'Water'

# 获取此状态下饱和液态水的密度
density_l = CP.PropsSI('D', 'T', temperature, 'P', pressure, fluid)
print(f"液态水的密度: {density_l:.2f} kg/m^3")

# 获取此状态下饱和蒸汽的密度
density_v = CP.PropsSI('D', 'T', temperature, 'Q', 1, fluid) # Q=1 表示饱和蒸汽
print(f"蒸汽的密度: {density_v:.4f} kg/m^3")

# 获取饱和液态水的比焓 (能量)
enthalpy_l = CP.PropsSI('H', 'T', temperature, 'P', pressure, fluid)
print(f"液态水的焓: {enthalpy_l/1000:.2f} kJ/kg")

# 获取饱和蒸汽的比焓
enthalpy_v = CP.PropsSI('H', 'T', temperature, 'Q', 1, fluid)
print(f"蒸汽的焓: {enthalpy_v/1000:.2f} kJ/kg")

# 计算汽化潜热
latent_heat = enthalpy_v - enthalpy_l
print(f"汽化潜热: {latent_heat/1000:.2f} kJ/kg")
```
